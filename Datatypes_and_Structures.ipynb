1. What are data structures, and why are they important ?

Data structures are ways of organizing and storing data in a computer so that it can be accessed and used efficiently. They define how data is related and what operations can be performed on it.

Importance of Data structures:

 **Memory Management**: They use memory in an organized way, avoiding wastage.
 **Code Organization**: Proper use of data structures makes programs easier to read, maintain, and scale.
 **Efficiency**: They help in performing tasks like searching, inserting, and deleting data quickly
 **Problem Solving**: Different problems require different structures, like queues for task scheduling or trees for hierarchical data.

2. Explain the difference between mutable and immutable data types with examples.

Data types in programming can be mutable or immutable depending on whether their value can be changed after creation.

Mutable: These are data types whose values can be modified after they are created. For example, lists in Python are mutable. You can change, add, or remove elements in a list without creating a new one.

Immutable: These are data types whose values cannot be changed once they are created. Any change creates a new object instead of modifying the original. Examples include strings and tuples.

3. What are the main differences between lists and tuples in Python ?

**Mutability**:

List: Mutable - you can change, add, or remove elements.
Example: my_list = [1, 2, 3]; my_list.append(4) → [1, 2, 3, 4]

Tuple: Immutable - once created, you cannot change its elements.
Example: my_tuple = (1, 2, 3) → cannot add or remove elements

**Syntax**:

List: Defined with square brackets [ ]

Tuple: Defined with parentheses ( )

**Performance**:

List: Slightly slower because of mutability

Tuple: Faster and more memory-efficient

**Usage**:

List: Use when data needs to change frequently

Tuple: Use when data should remain constant

4. Describe how dictionaries store data.

A dictionary is a data structure that stores data in the form of key-value pairs. Each key is unique and is used to access its corresponding value. Unlike lists or tuples, dictionaries are unordered, meaning the items do not have a fixed position.

Each item in a dictionary is stored as a pair: a key and its value. Keys act like addresses, allowing you to quickly retrieve the associated value. Dictionaries use a method called hashing to store and look up data efficiently, which makes searching very fast even for large datasets.

5. Why might you use a set instead of a list in Python ?

A set is used when you need to store unique items without duplicates. Unlike lists, sets do not maintain order, but they are much faster for checking membership (e.g., whether an element exists or not).

No Duplicates: Automatically removes repeated values.

Faster Lookup: Checking if an item exists in a set is quicker than in a list.

Useful for Operations: Sets support mathematical operations like union, intersection, and difference.

8. What is a hash table, and how does it relate to dictionaries in Python ?

A hash table is a data structure that stores data in key-value pairs. It uses a process called hashing, where a key is converted into a unique index to store and quickly retrieve the value. This makes operations like search, insert, and delete very efficient.

In Python, a dictionary is implemented using a hash table. Each key in a dictionary is hashed to find its location in memory, and the value is stored at that position. This is why dictionary lookups are very fast.

In dictionaries, the in keyword checks if a value exists as a key in the dictionary, not as a value.

17. Describe the time complexity of accessing elements in a dictionary.

In Python, dictionaries are implemented using hash tables, which allow very fast access to elements. The average time complexity for accessing a value by its key is O(1), meaning it takes constant time regardless of the number of items in the dictionary. However, in rare cases of hash collisions, the time may increase slightly, but Python handles collisions efficiently to keep access fast.

19. Why are dictionaries considered unordered, and how does that affect data retrieval ?

Dictionaries in Python are considered unordered because their items are stored based on hash values rather than the order they are added. This means you cannot access elements by position like in a list. However, data retrieval using a key is still very fast because dictionaries use hash tables, so you can efficiently get the value of any key even though the items have no fixed order.

20. Explain the difference between a list and a dictionary in terms of data retrieval.

The main difference between a list and a dictionary in terms of data retrieval is how elements are accessed. In a list, items are accessed by their index position, so you need to know the position of the element, and searching for a value can take longer (O(n) in the worst case). In a dictionary, items are accessed using keys, which allows for fast retrieval in constant time (O(1)) because dictionaries are implemented using hash tables.
